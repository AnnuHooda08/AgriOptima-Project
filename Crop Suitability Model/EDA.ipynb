```python
# ============================================
# STEP 2: DATA CLEANING / DATA QUALITY CHECK
# ============================================

# 1. Check missing values
print("Missing values:")
print(df.isnull().sum())

# 2. Check duplicate rows
print("\nDuplicate rows:")
print(df.duplicated().sum())

# 3. Check data types
print("\nData types:")
print(df.dtypes)

# 4. Check for negative values in numerical columns
numeric_cols = ['N', 'P', 'K', 'temperature', 'humidity', 'ph', 'rainfall']

print("\nNegative values:")
for col in numeric_cols:
    print(f"{col}: {(df[col] < 0).sum()}")

# 5. Check ranges of numerical features
print("\nMinimum values:")
print(df[numeric_cols].min())

print("\nMaximum values:")
print(df[numeric_cols].max())

# 6. Check duplicate rows in detail
duplicates = df[df.duplicated(keep=False)]

print("\nDuplicate rows:")
display(duplicates.head(20))

# 7. Check target labels
print("\nCrop labels:")
print(sorted(df['label'].unique()))

# 8. Final dataset information
print("\nFinal dataset information:")
print(df.info())
```

```python
# ============================================
# STEP 3: PREPARE DATA FOR ML
# ============================================

from sklearn.model_selection import train_test_split

# 1. Define features and target
X = df[['N', 'P', 'K', 'temperature', 'humidity', 'ph', 'rainfall']]
y = df['label']

print("Features (X):")
print(X.columns.tolist())

print("\nTarget (y):")
print(y.name)

print("\nX shape:", X.shape)
print("y shape:", y.shape)

# 2. Train-test split
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

# 3. Check split
print("\nTRAINING DATA")
print("X_train:", X_train.shape)
print("y_train:", y_train.shape)

print("\nTESTING DATA")
print("X_test:", X_test.shape)
print("y_test:", y_test.shape)

# 4. Check class distribution
print("\nTraining class distribution:")
print(y_train.value_counts())

print("\nTesting class distribution:")
print(y_test.value_counts())
```

```python
# ============================================
# STEP 4: TRAIN 5 CLASSIFICATION MODELS
# ============================================

from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier

# --------------------------------------------
# 1. Create the models
# --------------------------------------------

models = {

    "Logistic Regression": Pipeline([
        ("scaler", StandardScaler()),
        ("model", LogisticRegression(max_iter=2000))
    ]),

    "KNN": Pipeline([
        ("scaler", StandardScaler()),
        ("model", KNeighborsClassifier(n_neighbors=5))
    ]),

    "SVM": Pipeline([
        ("scaler", StandardScaler()),
        ("model", SVC())
    ]),

    "Decision Tree": Pipeline([
        ("model", DecisionTreeClassifier(random_state=42))
    ]),

    "Random Forest": Pipeline([
        ("model", RandomForestClassifier(
            n_estimators=100,
            random_state=42
        ))
    ])
}

# --------------------------------------------
# 2. Train all 5 models
# --------------------------------------------

for name, model in models.items():

    model.fit(X_train, y_train)

    print(f"{name} trained successfully.")
```

```python
# ============================================
# STEP 5: MODEL EVALUATION
# ============================================

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report
)

results = []

for name, model in models.items():

    # Predict on unseen test data
    y_pred = model.predict(X_test)

    # Calculate metrics
    accuracy = accuracy_score(y_test, y_pred)
    precision = precision_score(
        y_test, y_pred, average='weighted', zero_division=0
    )
    recall = recall_score(
        y_test, y_pred, average='weighted', zero_division=0
    )
    f1 = f1_score(
        y_test, y_pred, average='weighted', zero_division=0
    )

    results.append({
        "Model": name,
        "Accuracy": accuracy,
        "Precision": precision,
        "Recall": recall,
        "F1 Score": f1
    })

# Convert results into DataFrame
results_df = pd.DataFrame(results)

# Sort by F1 Score
results_df = results_df.sort_values(
    by="F1 Score",
    ascending=False
).reset_index(drop=True)

# Display results
display(results_df)
```

```python
# Detailed classification report for each model

for name, model in models.items():

    y_pred = model.predict(X_test)

    print("\n" + "=" * 70)
    print(name)
    print("=" * 70)

    print(classification_report(
        y_test,
        y_pred,
        zero_division=0
    ))
```

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
import matplotlib.pyplot as plt

# Best model
best_model = models["SVM"]

# Predictions
y_pred_svm = best_model.predict(X_test)

# Confusion matrix
cm = confusion_matrix(y_test, y_pred_svm, labels=best_model.classes_)

plt.figure(figsize=(14, 12))

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=best_model.classes_
)

disp.plot(
    xticks_rotation=90,
    cmap="Blues",
    values_format="d"
)

plt.title("SVM - Confusion Matrix")
plt.tight_layout()
plt.show()
```

```python
from sklearn.metrics import classification_report

print(classification_report(
    y_test,
    y_pred_svm,
    zero_division=0
))
```

```python
# ============================================
# STEP 6: SVM HYPERPARAMETER TUNING
# ============================================

from sklearn.model_selection import GridSearchCV
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC

# Create SVM pipeline
svm_pipeline = Pipeline([
    ("scaler", StandardScaler()),
    ("svm", SVC())
])

# Parameters to test
param_grid = {
    "svm__C": [0.1, 1, 10, 100],
    "svm__kernel": ["linear", "rbf", "poly"],
    "svm__gamma": ["scale", "auto"]
}

# Grid Search
grid_search = GridSearchCV(
    estimator=svm_pipeline,
    param_grid=param_grid,
    cv=5,
    scoring="f1_weighted",
    n_jobs=-1,
    verbose=1
)

# Train and search
grid_search.fit(X_train, y_train)

# Best parameters
print("\nBest Parameters:")
print(grid_search.best_params_)

# Best cross-validation score
print("\nBest Cross-Validation F1 Score:")
print(grid_search.best_score_)
```

```python
# ============================================
# STEP 7: EVALUATE TUNED SVM ON TEST DATA
# ============================================

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report
)

# Best tuned model
best_svm = grid_search.best_estimator_

# Predict on unseen test data
y_pred_tuned = best_svm.predict(X_test)

# Calculate metrics
accuracy = accuracy_score(y_test, y_pred_tuned)
precision = precision_score(
    y_test, y_pred_tuned,
    average="weighted",
    zero_division=0
)
recall = recall_score(
    y_test, y_pred_tuned,
    average="weighted",
    zero_division=0
)
f1 = f1_score(
    y_test, y_pred_tuned,
    average="weighted",
    zero_division=0
)

print("TUNED SVM RESULTS")
print("=" * 40)
print(f"Accuracy  : {accuracy:.4f}")
print(f"Precision : {precision:.4f}")
print(f"Recall    : {recall:.4f}")
print(f"F1 Score  : {f1:.4f}")

print("\nClassification Report:")
print(classification_report(
    y_test,
    y_pred_tuned,
    zero_division=0
))
```
